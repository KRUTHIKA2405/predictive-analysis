# Regression Tree Implementation from Scratch

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Generate a simple regression dataset
np.random.seed(42)
X = np.sort(np.random.rand(50) * 10)
y = np.sin(X) + np.random.randn(50) * 0.5

regression_df = pd.DataFrame({'X': X, 'y': y})

print("Our simple regression dataset:")
display(regression_df.head())

plt.figure(figsize=(8, 6))
sns.scatterplot(x='X', y='y', data=regression_df)
plt.title('Simple Regression Dataset')
plt.xlabel('Feature (X)')
plt.ylabel('Target (y)')
plt.grid(True)
plt.show()
```

Now, let's define the impurity metric for regression trees: Mean Squared Error (MSE).

```python
def calculate_mse(y_values):
    """Calculates the Mean Squared Error for a set of target values."""
    if len(y_values) == 0:
        return 0
    mean_y = np.mean(y_values)
    mse = np.mean((y_values - mean_y)**2)
    return mse

print("MSE calculation function defined.")
```

Next, we need a function to find the best possible split point for a given node. This involves iterating through all features and all unique values within each feature to find the split that minimizes the weighted MSE of the child nodes.

```python
def find_best_split(data):
    """Finds the best feature and split value to minimize impurity (MSE)."""
    best_mse = float('inf')
    best_feature = None
    best_value = None

    n_samples = len(data)
    if n_samples <= 1: # Cannot split if less than 2 samples
        return None, None, None

    features = [col for col in data.columns if col != 'y']

    for feature in features:
        unique_values = data[feature].unique()
        for value in unique_values:
            # Split data based on the current feature and value
            left_split = data[data[feature] <= value]
            right_split = data[data[feature] > value]

            # Calculate weighted MSE
            mse_left = calculate_mse(left_split['y'].values)
            mse_right = calculate_mse(right_split['y'].values)

            # Avoid division by zero if a split results in an empty child
            if len(left_split) == 0 or len(right_split) == 0:
                continue

            weighted_mse = (len(left_split) / n_samples * mse_left) + \
                           (len(right_split) / n_samples * mse_right)

            if weighted_mse < best_mse:
                best_mse = weighted_mse
                best_feature = feature
                best_value = value

    return best_feature, best_value, best_mse

print("Best split finding function defined.")
```

Now, we will build the regression tree recursively. Each node will either be a split node (containing a feature and split value) or a leaf node (containing the predicted value, which is the mean of the target values in that node).

```python
def build_tree(data, max_depth, min_samples_split, current_depth=0):
    """Recursively builds the regression tree."""
    # Base cases for stopping the recursion
    if current_depth >= max_depth or len(data) < min_samples_split or calculate_mse(data['y'].values) == 0:
        # This is a leaf node, return the average of target values
        return {'type': 'leaf', 'prediction': np.mean(data['y'].values)}

    feature, value, mse = find_best_split(data)

    # If no good split found, make it a leaf node
    if feature is None:
        return {'type': 'leaf', 'prediction': np.mean(data['y'].values)}

    # Create child nodes
    left_split = data[data[feature] <= value]
    right_split = data[data[feature] > value]

    # Handle cases where a split creates an empty child node (should not happen if find_best_split is robust)
    if len(left_split) == 0 or len(right_split) == 0:
        return {'type': 'leaf', 'prediction': np.mean(data['y'].values)}

    node = {
        'type': 'split',
        'feature': feature,
        'value': value,
        'left': build_tree(left_split, max_depth, min_samples_split, current_depth + 1),
        'right': build_tree(right_split, max_depth, min_samples_split, current_depth + 1)
    }
    return node

print("Tree building function defined.")
```

Finally, let's implement the prediction function, which traverses the tree for a given new observation.

```python
def predict_observation(tree, observation):
    """Predicts the target value for a single observation using the trained tree."""
    if tree['type'] == 'leaf':
        return tree['prediction']

    feature = tree['feature']
    value = tree['value']

    if observation[feature] <= value:
        return predict_observation(tree['left'], observation)
    else:
        return predict_observation(tree['right'], observation)

print("Prediction function defined.")
```

Let's put it all together and train our simple regression tree, then make some predictions and visualize them.

```python
# Train the Regression Tree
max_depth = 3
min_samples_split = 5

print(f"Building tree with max_depth={max_depth} and min_samples_split={min_samples_split}...")
regression_tree = build_tree(regression_df, max_depth, min_samples_split)
print("Tree built successfully.")

# Make predictions on the training data to see how well it fits
regression_df['y_predicted'] = regression_df.apply(lambda row: predict_observation(regression_tree, row), axis=1)

# Predict for a new unseen point
new_observation = pd.Series({'X': 4.5})
predicted_value_new_point = predict_observation(regression_tree, new_observation)

print(f"\nPrediction for new observation X={new_observation['X']}: {predicted_value_new_point:.2f}")

# Visualize the results
plt.figure(figsize=(10, 7))
sns.scatterplot(x='X', y='y', data=regression_df, label='Actual Data')
sns.lineplot(x='X', y='y_predicted', data=regression_df, color='red', linestyle='--', label='Tree Predictions')
plt.scatter(new_observation['X'], predicted_value_new_point, color='green', marker='o', s=200, label=f'New Point Prediction (X={new_observation['X']:.1f}, y={predicted_value_new_point:.2f})')
plt.title(f'Regression Tree Predictions (Max Depth: {max_depth})')
plt.xlabel('Feature (X)')
plt.ylabel('Target (y)')
plt.legend()
plt.grid(True)
plt.show()
```
